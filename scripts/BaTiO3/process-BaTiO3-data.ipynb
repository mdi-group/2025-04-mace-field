```python
from ase.io import read, write
import numpy as np
```

```python
BaTiO3_atoms = read("BaTiO3.xyz", ":")
```

```python
for atom in BaTiO3_atoms:
    forces = atom.calc.__dict__['results']['forces']
    stress = atom.calc.__dict__['results']['stress']

    polarisation = atom.calc.__dict__['results']['polarization']
    becs = atom.arrays.pop('born_charge')
    polarisability = atom.info.pop("polarizability")

    volume = atom.get_volume()

    atom.info['REF_energy'] = atom.info.pop('total_energy')
    atom.arrays['REF_forces'] = forces
    atom.info['REF_stress'] = stress

    atom.info['REF_polarization'] = polarisation / volume
    atom.arrays['REF_becs'] = becs
    eps0 = 8.8541878128e-12 / 1.602176634e-19 / 1e10
    atom.info['REF_polarizability'] = polarisability / volume / eps0
    atom.info['REF_electric_field'] = np.array([0.0, 0.0, 0.0])
    atom.calc = None
```

```python
write("BaTiO3-preprocessed.xyz", BaTiO3_atoms)
```
